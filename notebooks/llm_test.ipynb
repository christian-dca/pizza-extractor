```python
from dotenv import load_dotenv
import os
import getpass

load_dotenv()

if "AWS_BEARER_TOKEN_BEDROCK" not in os.environ:
    os.environ["AWS_BEARER_TOKEN_BEDROCK"] = getpass.getpass("Introduce tu  AWS Key: ")

if "ANTHROPIC_API_KEY" not in os.environ:
    os.environ["ANTHROPIC_API_KEY"] = getpass.getpass("Introduce tu  AWS Key: ")
```

```python
import os
from langchain.chat_models import init_chat_model

model = init_chat_model(
    "eu.amazon.nova-2-lite-v1:0",
    model_provider="bedrock_converse",
    max_tokens = 500
)
```

```python
response = model.invoke("Hola")
response
```
